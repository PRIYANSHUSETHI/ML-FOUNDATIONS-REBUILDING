WORKING WITH DATES

```python
# first we convert the data type of the date column to date-time datatype
date['date'] = pd.to_datetime(date['date'])
```

```python
# extracting year from the date
date["date_year"] = date['date'].dt.year

date.head()
```

```python
# extracting the month
date['date_month'] = date['date'].dt.month
date.head()
```

```python
# placing name of the month in the dataframe
date['date_month_name'] = date['date'].dt.month_name()
date.head()
```

```python
# extracting the day of the month
date["date_day"] = date['date'].dt.day
date.head()
```

```python
# finding day of the week
date['day_of_week'] = date['date'].dt.day_of_week
date.head()
```

```python
# finding name of the day of the week
date['day_of _the_week_name'] = date['date'].dt.day_name()
date.head()
```

```python
# is it a weekend on the day
date['is_weekend'] = np.where(date['day_of _the_week_name'].isin(['Sunday','Saturday']),1,0) # is the day matches sat or sunday then 1 otherwise 0
date.head()
```

```python
# extracting week of the year
date['date_week'] = date['date'].dt.isocalendar().week
date.head()
```

```python
# extracting quarter
date['quarter'] = date['date'].dt.quarter
date.head()
```

```python
# if we want we can also see the semester of the date
date['semester'] = np.where(date['quarter'].isin([1,2]), 1, 2)
date.head()
```

```python
today - date['date']
```

```python
# days passed from that day to today
(today - date['date']).dt.days
```

```python
# months passed
np.round((today - date['date']) / pd.Timedelta(days=30.44), 0)
```

WORKING WITH TIME VARIABLES

```python
# converting the date column in time dataframe to date-time data type
time['date'] = pd.to_datetime(time['date'])
```

```python
# extracting hours, minutes and seconds
time['hour'] = time['date'].dt.hour
time['min'] = time['date'].dt.minute
time['sec'] = time['date'].dt.second

time.head()
```

```python
# extracting only the time
time['date_time'] = time['date'].dt.time
time.head()
```
